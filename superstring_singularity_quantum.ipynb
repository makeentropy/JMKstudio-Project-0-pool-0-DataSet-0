## 6. 拓扑维度空间搜索

```python
from topology import TopologySearchEngine

# 初始化拓扑搜索引擎（11维空间）
topo_engine = TopologySearchEngine(dimensions=11)
print("拓扑搜索引擎初始化完成 (11维空间)")
```

```python
# 添加示例节点
for i in range(20):
    node = topo_engine.add_node(
        node_id=f"node_{i:02d}",
        data={"type": "quantum", "index": i},
        dimensions=np.random.randn(11).tolist()
    )
    topo_engine.add_metadata(f"node_{i:02d}", "group", i % 5)

print("已添加20个量子节点到拓扑空间")
```

```python
# 连接节点
for i in range(19):
    topo_engine.connect_nodes(f"node_{i:02d}", f"node_{i+1:02d}", weight=0.8)

print("节点连接完成")
```

```python
# 向量相似度搜索
query_vector = np.random.randn(11).tolist()
results = topo_engine.search_by_vector(query_vector, top_k=5)

print("向量相似度搜索结果 (Top 5):")
for result in results:
    print(f"  {result['node_id']}: 相似度 = {result['score']:.4f}")
```

```python
# 按元数据搜索
group_nodes = topo_engine.search_by_metadata("group", 2)
print(f"Group 2中的节点: {[n.node_id for n in group_nodes]}")
```
